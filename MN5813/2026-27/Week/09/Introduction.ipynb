## The room

In each crit, one group presents one of its figures. All the other groups visit and critique it.

Roles in every crit:

- **Critics**: Every visiting group. Everyone speaks in every round: Describe, Decode, Diagnose, and Redesign. Each visiting group fills in its own feedback sheet
- **Authors**: The whole presenting group. Watch which parts of your figure the critics struggle to read

I keep time and call the start of each round.

## The Four-D protocol

The [DataViz design guide](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/dataviz-design-guide.md) defines each round in section "8. The Four-D crit protocol".

- **Round 1, Describe**
- **Round 2, Decode**
- **Round 3, Diagnose**: For each [TRACE](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-trace) letter, write an "I like" and an "I wish" on the [feedback sheet](#the-feedback-sheet). At **E**, turn over the authors' rationale and read it
- **Round 4, Redesign**: Each critic asks one "What if…?" question
- **Author response**: The authors say which change they will make, **with a reason**. Each visiting group writes it on its sheet. Then each group hands its sheet to the authors

<a id="the-feedback-sheet"></a>

## The feedback sheet

Each visiting group fills in one [feedback sheet](./feedback-sheet.pdf) for each figure it critiques. I bring printed copies. The sheet has a part for each round:

- **Describe**: What you see, facts only
- **Decode**: The figure's message, in one sentence from each critic
- **Diagnose**: An "I like" and an "I wish" for each TRACE letter, and the biggest problem
- **Redesign**: One "What if…?" question from each critic, and the change the authors will make

A separate page of the sheet explains TRACE and the Four-D crit. Your group keeps each sheet it receives.

- Count each accepted change as one design <!-- plain-language: everyday intent -->intent, however many code edits it takes. Fix a mechanical "I wish", such as a missing unit or source line, as a routine correction outside that change. Log any repair that changes what the reader concludes, such as a new takeaway title, inside the `DECISION:` entry
