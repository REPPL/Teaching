## The setup: Group A's brief

Group A's report section asks: **"Has Team GB closed the gap to the leading nation since 1996?"** This is a change-over-time question, for a non-technical reader of a printed report. The cell below generates their synthetic working data. It contains total medals per Summer Games for six nations, 1988–2024.

````python
# --- Data setup: Run this cell first (runs anywhere, no download needed) ---
# Each nation's trend below is invented. Seeded noise is added to it.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(5813)  # module seed: Do not change

games = np.arange(1988, 2028, 4)   # ten Summer Games, 1988-2024
t = np.arange(len(games))

trend = {
    "United States": 100 + 1.5 * t,
    "China":         28 + 60 / (1 + np.exp(-(t - 4))),
    "Russia":        92 - 5.5 * t,
    "Germany":       70 - 4.0 * t,
    "Great Britain": 22 + np.where(t > 2, (t - 2) * 7.0, 0),
    "Australia":     24 + 22 * np.exp(-((t - 3) ** 2) / 3),
}
medals = pd.DataFrame(
    {nation: np.rint(values + rng.normal(0, 4, len(t))).clip(min=8).astype(int)
     for nation, values in trend.items()},
    index=pd.Index(games, name="games_year"),
)

print(f"Created medals: {medals.shape[0]} Games x {medals.shape[1]} nations "
      "(synthetic, module seed 5813)")
medals
````

````python
# Exhibit A: The Associate's first draft.
fig, ax = plt.subplots(figsize=(8, 4.5))
for nation in medals.columns:
    ax.plot(medals.index, medals[nation], label=nation)
ax.set_title("Medal Counts by Country")
ax.set_ylabel("value")
ax.legend(loc="upper right")
ax.grid(True)
plt.show()
````

## Exhibit B: The redesign

Sam directed the Associate to redraw the figure. This is the [GENERATE-JUDGE](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-generate-judge) move, with the crit record as the <!-- plain-language: everyday brief -->brief. Sam's prompt reads:

> *"Redesign the figure per our crit record: Grey all nations except Great Britain and the United States. Show Great Britain in Okabe-Ito blue #0072B2 and the leader in dark grey. Use no legend. Label every line directly at its right-hand end, nudging labels apart so none overlap. Turn the top and right spines off. Use light horizontal gridlines only. Label the y-axis 'Total medals per Games'. Put the x-ticks on Games years. Add a takeaway title stating the gap has roughly halved since 1996, with a descriptive subtitle. Add a source line naming the synthetic data and seed."*

Check every claim before you put it in a title. The new title will claim the gap has "roughly halved", and the cell below checks that.

````python
gap = medals["United States"] - medals["Great Britain"]
print(f"Gap to leader in 1996: {gap.loc[1996]} medals")
print(f"Gap to leader in 2024: {gap.loc[2024]} medals")
print(f"2024 gap as a share of the 1996 gap: {gap.loc[2024] / gap.loc[1996]:.0%}")
````

The output shows the 2024 gap is about half the 1996 gap, so "roughly halved" is accurate. The cell below is the code the prompt produced, after Sam had checked it. Two pieces of it are new. A small function, `spread_labels`, moves the six labels apart so none overlap. [`subplots_adjust`](https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.subplots_adjust.html) narrows the plot to leave room for the labels at the right-hand edge.

Inside `spread_labels`, `sorted` with `key=lambda kv: -kv[1]` puts the labels in order from the highest value down (Optional background: [Week 2b](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Week/02b/Introduction.ipynb)).

````python
# Exhibit B: The human-directed revision.
ACCENT, LEADER, PACK = "#0072B2", "#444444", "#b8b8b8"


def spread_labels(last_values, min_gap=7):
    """Nudge direct-label y-positions apart (top to bottom) so none overlap."""
    positions, previous = {}, None
    for name, value in sorted(last_values.items(), key=lambda kv: -kv[1]):
        y = value if previous is None else min(value, previous - min_gap)
        positions[name] = y
        previous = y
    return positions


label_y = spread_labels(medals.iloc[-1].to_dict())

fig_redesign, ax = plt.subplots(figsize=(8, 4.5))

for nation in medals.columns:                      # the other nations, in grey
    if nation in ("Great Britain", "United States"):
        continue
    ax.plot(medals.index, medals[nation], color=PACK, linewidth=1.2)
    ax.text(2025.5, label_y[nation], nation, color="#888888",
            fontsize=9, va="center")

ax.plot(medals.index, medals["United States"], color=LEADER, linewidth=1.8)
ax.text(2025.5, label_y["United States"], "United States (leader)",
        color=LEADER, fontsize=10, va="center")

ax.plot(medals.index, medals["Great Britain"], color=ACCENT, linewidth=2.5)
ax.text(2025.5, label_y["Great Britain"], "Great Britain",
        color=ACCENT, fontsize=10, fontweight="bold", va="center")

ax.spines[["top", "right"]].set_visible(False)
ax.set_xticks([1988, 1996, 2004, 2012, 2024])
ax.set_yticks([0, 40, 80, 120])
ax.set_ylim(0, 130)
ax.grid(axis="y", color="#e6e6e6", linewidth=0.8)
ax.set_axisbelow(True)
ax.set_ylabel("Total medals per Games")

ax.set_title("Team GB has roughly halved the gap to the leader since 1996",
             loc="left", fontsize=13, fontweight="bold", pad=22)
ax.text(0, 1.03, "Total medals per Summer Games, six leading nations, 1988–2024",
        transform=ax.transAxes, fontsize=10, color="#555555")
fig_redesign.text(0.01, -0.05,
                  "Source: Synthetic demonstration data (module seed 5813).",
                  fontsize=8, color="#777777")
fig_redesign.subplots_adjust(right=0.76)
plt.show()
````

## Preparing your figures: Before the Gallery, and for the report

Two candidate figures exported at final size are part of what you bring to the Gallery ([Introduction.ipynb](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Week/09/Introduction.ipynb), section "Preparation").

### Export quality

- **Design at final size.** Set `figsize` to the size the figure will occupy on the page (a full-width report figure is roughly 8 x 4.5 inches). Judge text at that size, as the [design guide](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/dataviz-design-guide.md), section "4. A: Accessible", explains. If you scale the image in your document afterwards, you undo every font-size <!-- plain-language: everyday decision -->decision you made
- **Choose the file format.** Use PNG at `dpi=200` (dots per inch) or more for Word and Google Docs. Use PDF or SVG (sharp at any size) where your report tool accepts it. Both come from the same [`savefig`](https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.savefig.html) call, and the file extension picks the format
- **Export with `bbox_inches="tight"`.** It trims empty margins and keeps elements outside the axes (direct labels, source lines) in the export
- **Never screenshot a figure.** Screenshots blur text. An image with no code behind it cannot be regenerated when the data changes. Make every figure in your report reproducible by running a cell
- **Name exports so they sort in order.** Number them to match the report, and add a few hyphenated words so the file is findable: `fig01-gap-to-leader.png`, `fig02-...`

The cell below exports Exhibit B as PNG and PDF and lists the files.

````python
import os

FIG_DIR = "assets/figures/week09"
os.makedirs(FIG_DIR, exist_ok=True)

fig_redesign.savefig(f"{FIG_DIR}/fig01-gap-to-leader.png",
                     dpi=200, bbox_inches="tight")
fig_redesign.savefig(f"{FIG_DIR}/fig01-gap-to-leader.pdf",
                     bbox_inches="tight")

for name in sorted(os.listdir(FIG_DIR)):
    size_kb = os.path.getsize(os.path.join(FIG_DIR, name)) / 1024
    print(f"Exported {FIG_DIR}/{name} ({size_kb:.0f} KB)")
````

### Consistent styling across a report

Figures with mixed fonts, palettes, and source-line formats look like separate pieces of work. To keep them consistent, put the same styling cell at the top of every figure notebook your group owns. Agree the cell once and log it as a `DECISION:`. The cell updates [`rcParams`](https://matplotlib.org/stable/users/explain/customizing.html), matplotlib's dictionary of defaults, so every figure drawn after it uses the house style:

````python
# House style: Copy this cell to the top of every figure notebook.
# Change it only through a team DECISION: entry.
OKABE_ITO = ["#0072B2", "#E69F00", "#009E73", "#D55E00",
             "#56B4E9", "#CC79A7", "#F0E442", "#000000"]

plt.rcParams.update({
    "figure.figsize": (8, 4.5),
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.titleweight": "bold",
    "axes.titlelocation": "left",
    "axes.prop_cycle": plt.cycler(color=OKABE_ITO),
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.grid.axis": "y",
    "grid.color": "#e6e6e6",
    "grid.linewidth": 0.8,
    "axes.axisbelow": True,
})
print("House style applied.")
````

The cell below sets no styling of its own. The figure it draws gets its fonts, colours, spines, and gridlines from the house style cell.

````python
fig_check, ax = plt.subplots()
ax.plot(medals.index, medals["Great Britain"], linewidth=2)
ax.set_ylabel("Total medals per Games")
ax.set_title("Styled by the house style alone")
plt.show()
````
