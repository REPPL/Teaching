## The presentation format

Each team gets **fifteen minutes**, in the running order I post at the start of the session.

- **Minutes 0–10:** Your team presents its questions, its main findings, and the <!-- plain-language: everyday decision -->decisions behind them
- **Minutes 10–15:** I ask **"walk me through this <!-- plain-language: everyday decision -->decision"** questions. I draw them from the [`DECISION:` entries](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-decision-entry) in your team's record and from your Prompt Diaries

The questions are the ones a client or a marker asks before trusting the work. For example:

The cells below contain this week's code (the little there is). They show one process indicator on a mock <!-- plain-language: everyday decision -->decision log, and take about thirty seconds. Run them. Then repeat the check on your own team's log before you submit. To do that, replace the mock *roster* and *entries* in the setup cell with your team's own.

````python
# --- Data setup: Run this cell first (runs anywhere, no download needed) ---
# A mock team decision log. Retype your own log from your team's exported record, in the same layout.
import pandas as pd

# The team as registered, including anyone who has yet to log a decision.
roster = ["Priya", "Dan", "Mo", "Sofia", "Ben"]

decisions = pd.DataFrame({
    "week":   [4, 5, 5, 6, 7, 8, 9, 9, 10],
    "member": ["Priya", "Dan", "Priya", "Mo", "Sofia", "Dan", "Mo", "Sofia", "Priya"],
    "decision": [
        "Rotate weekly export duty, starting with Dan",
        "Treat an empty attendance time as 'no engine attended', not missing data",
        "Count engine minutes per incident, not per mobilisation row",
        "Scope the project to 2 questions on response times and 1 on false alarms",
        "Join on IncidentNumber only, never on borough names",
        "Keep unmatched mobilisations with a left join, and flag them",
        "Replace the stacked bar chart (Fig 3) with a slope chart, after the crit",
        "Use the Okabe-Ito colour palette in every figure",
        "Freeze the dataset; no cleaning changes before submission",
    ],
})
print(f"Created mock decision log in memory: {len(decisions)} DECISION: entries")
````

The next cell computes (calculates) the indicator: The number of <!-- plain-language: everyday decision -->decisions per member, and the weeks the log covers. It uses one pandas method you have not met before. [`reindex`](https://pandas.pydata.org/docs/reference/api/pandas.Series.reindex.html) lines the counts up against the full roster, so a member with no entries still appears, with zero.

````python
print(decisions['member'].value_counts().reindex(roster, fill_value=0))
print(f"\nWeeks with at least one logged decision: {sorted(decisions['week'].unique().tolist())}")
````

Notice Ben in the output. He is on the roster but has no entries in the log. If your log shows an uneven spread, I ask about it at the presentation. Be ready to explain it. Check your own log today for a member with no entries, or for three weeks in a row with none.
