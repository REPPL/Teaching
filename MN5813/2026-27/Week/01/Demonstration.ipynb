## 1. Numbers and text

Every Python value has a *type*. You use three types constantly:

- **int**: A whole number (`42`)
- **float**: A number with a decimal point (`3.50`)
- **str** (string): Text, wrapped in quotes (`"flat white"`)

`3.50` is a number you can do arithmetic with. `"3.50"` is text that only looks like a number. A spreadsheet column of numbers often arrives as text. The `type()` function tells you which one you are holding. [Python's official tutorial](https://docs.python.org/3/tutorial/introduction.html) also covers numbers and text.

The next cell prints the type of four values.

````python
print(type(42))       # int: a whole number
print(type(3.50))     # float: a decimal number
print(type("3.50"))   # str: text that LOOKS like a number
print(type("flat white"))  # str: ordinary text
````

## 2. Variables: Labelled boxes

A **variable** is a labelled box that holds a value, so you can use the value later by name. The `=` sign puts the value in the box. Read it as "becomes", not "equals". The next cell stores three values and uses them by name.

````python
price = 3.50          # put 3.50 in a box labelled 'price'
drink = "flat white"  # put text in a box labelled 'drink'
cups = 2

total = price * cups   # use the boxes by name
print(drink, "x", cups, "costs", total)
````

Three things to notice:

- You choose the names. `price` is clearer than `p`. Python's convention is lowercase words joined by underscores (`cups_sold`)
- `#` starts a **comment**: A note for humans that Python ignores
- Putting a new value in the box replaces the old one. Run `cups = 3` and the old `2` is gone

## 3. f-strings: Readable output

An **f-string** (Python's [formatted string literal](https://docs.python.org/3/tutorial/inputoutput.html)) lets you write a sentence and drop variables into it with curly braces. Note the `f` before the opening quote. The next cell prints one sentence from three variables.

````python
drink = "flat white"
price = 3.50
cups = 2

print(f"{cups} x {drink} comes to £{price * cups}")
````

The output ends in `£7.0`. Money needs *two* decimal places. Put `:.2f` after the variable name, inside the braces. It means "format as a number with 2 decimal places". The next cell prints the same line with that fix.

````python
print(f"{cups} x {drink} comes to £{price * cups:.2f}")
print(f"Unit price: £{price:.2f}")   # :.2f = show 2 decimal places
````

## 4. Comparisons and decisions: if / else

Analysis is full of yes/no questions: Is this order over budget? Did sales fall? A **comparison** asks the question and answers `True` or `False`. The next cell asks three of these questions.

````python
budget = 4.00
price = 3.50

print(price > budget)    # is price over budget?
print(price <= budget)   # is it within budget?
print(price == 3.50)     # is it exactly 3.50?  (== compares; = assigns)
````

Two equals signs (`==`) *compare*. One equals sign (`=`) *assigns*. Python complains when you mix them up.

An **if statement** acts on the answer. The indented lines run only when the condition is `True`. The `else` block runs otherwise. **Indentation is grammar in Python.** In the next cell, the four spaces at the start of a line tell Python that the line belongs inside the `if`. The official tutorial's [control flow chapter](https://docs.python.org/3/tutorial/controlflow.html) covers `if` and the loops you meet in section 6.

````python
price = 3.50
budget = 4.00

if price > budget:
    print(f"£{price:.2f} is over budget. Flag it.")
else:
    print(f"£{price:.2f} is within the £{budget:.2f} budget.")
````

## 6. for loops: Do something with every item

A **for loop** visits each item in a list, one at a time, and runs the indented block for it. Read it aloud as "for each price in prices". Run the next cell to print each price.

````python
prices = [3.50, 2.20, 3.80, 2.40, 3.60]

for price in prices:
    print(f"item costs £{price:.2f}")
````

The loop variable (`price`) is a name you choose. It takes each value in turn. The classic analyst pattern is the **running total**: Start a counter at zero and add each item as you visit it. The next cell totals the price list this way.

````python
prices = [3.50, 2.20, 3.80, 2.40, 3.60]

total = 0
for price in prices:
    total = total + price   # add this price to the running total

print(f"Total: £{total:.2f}")
````

## 7. len, sum, min, max: The shortcuts

Python has **built-in functions** that answer everyday questions about a list. [The full table of built-in functions](https://docs.python.org/3/library/functions.html) is one page of the official docs. The next cell answers five questions about the price list.

````python
prices = [3.50, 2.20, 3.80, 2.40, 3.60]

print(f"How many:        {len(prices)}")
print(f"Total:           £{sum(prices):.2f}")
print(f"Cheapest:        £{min(prices):.2f}")
print(f"Most expensive:  £{max(prices):.2f}")
print(f"Average:         £{sum(prices) / len(prices):.2f}")
````

Python has no built-in `average()`. Analysts write `sum(...) / len(...)`, which is the total divided by the count. You meet it again in Quest 01.

## 8. Functions: Calling vs defining

You have been **calling** functions all along: `print(...)`, `type(...)`, `sum(...)`. You write the name, put the inputs in the brackets, and get the result back.

You can also **define** your own functions, to bundle a few steps under a name and reuse them. `def` starts the definition. `return` hands back the result. Nothing runs until you *call* it. The next cell defines one function and calls it twice.

````python
def describe_price(price, budget):
    """Return a one-line verdict on a price against a budget."""
    # ^ the text in triple quotes is a docstring: a note describing the
    #   function. Python skips over it, like a # comment.
    if price > budget:
        return f"£{price:.2f}: OVER the £{budget:.2f} budget"
    else:
        return f"£{price:.2f}: within budget"

# Defining did nothing visible. CALLING makes it work:
print(describe_price(3.50, 4.00))
print(describe_price(4.80, 4.00))
````

## 9. Reading an error message, top to bottom

The next cell makes a common analyst's mistake: It adds a *number* to a *price stored as text* (the `"3.50"` from section 1). Run it and **read the error before reading on**. The official tutorial has [a chapter on errors and exceptions](https://docs.python.org/3/tutorial/errors.html).

````python
# One of these 'numbers' is actually text: a very common data problem.
price_from_spreadsheet = "3.50"   # arrived as text!
service_charge = 0.30

total = price_from_spreadsheet + service_charge
print(total)
````

How to read any Python error, top to bottom:

1. **The arrow or highlighted line** shows *where* it went wrong: The `total = ...` line
2. **The last line names the problem**: `TypeError: can only concatenate str (not "float") to str`. Python is saying "you asked me to add text and a number, and I will not guess what you meant"
3. **Everything in between** is the route Python took to get there. Skim it

The fix is `float()`, which turns number-looking text into a number: `float("3.50")` gives `3.5`. The next cell runs the corrected version.

````python
price_from_spreadsheet = "3.50"
service_charge = 0.30

total = float(price_from_spreadsheet) + service_charge   # float() rescues the text
print(f"Total with service: £{total:.2f}")
````

## 10. Putting it together: The mini price report

The next cell does one small job: **Data in a list, then loop and compute, then report with f-strings**. Predict what it prints, then run it and check.

````python
# A tiny price list: what the office kitchen sells.
menu = [
    ("flat white", 3.50),
    ("espresso", 2.20),
    ("mocha", 3.80),
    ("tea", 2.40),
    ("hot chocolate", 3.60),
]

total = 0
over_three = 0

for name, price in menu:      # each item unpacks into a name and a price
    total = total + price
    if price > 3.00:
        over_three = over_three + 1

average = total / len(menu)

print(f"Menu report: {len(menu)} drinks")
print(f"Total if you bought one of each: £{total:.2f}")
print(f"Average price: £{average:.2f}")
print(f"Drinks over £3.00: {over_three} of {len(menu)}")
````

One new thing: Each menu item is a **pair** in round brackets (`("flat white", 3.50)`), called a *tuple*. The loop line `for name, price in menu:` unpacks each pair into two variables at once. You use this shape in Quest 01.
