````python
# --- Data setup: Run this cell first (runs anywhere, no download needed) ---
# Friday's coffee run from Beans of Egham, typed up from the receipt:
# 12 items of (item, price).
# No random generation needed here: The list is fixed as received.
# (From Week 2 onwards, data-setup cells use the module seed, 5813.)

coffee_run = [
    ("flat white", 3.40),
    ("latte", 3.60),
    ("cappuccino", 3.50),
    ("espresso", 2.20),
    ("hot chocolate", 3.20),
    ("english breakfast tea", 2.40),
    ("peppermint tea", 2.40),
    ("almond croissant", 2.95),
    ("filter coffee", "2.80"),
    ("banana bread", 2.75),
    ("mocha", 3.80),
    ("oat flat white", 3.80),
]

print(f"Created coffee_run: {len(coffee_run)} items from Friday's receipt")
````

## 🔍 First look: What the three cells were telling you

1. It prints `12`, then `('flat white', 3.4)`. Each record is a pair in round brackets: A name, then a price. `:.2f` looks odd, but it is really helpful for formatting numbers in text. It is essentially its own little programming language: The `.2` means two digits after the decimal point, and `f` means formatted as a float. Python drops the trailing zero of `3.40`. So every number that represents money is best printed with `:.2f`
2. It prints three lines: `flat white: £3.40`, `latte: £3.60`, and `cappuccino: £3.50`. Again, the `:.2f` in the f-string puts the zero back
3. It prints twelve lines, and they are not all the same. Eleven say `<class 'float'>`. The filter coffee says `<class 'str'>`: Its price is the text `"2.80"`

If you predicted twelve floats, you made the natural guess. That surprise (if it was one) is a helpful lesson: A value can look like a number and print like one, yet still be text.

The cell below is a habit worth keeping. It only prints the records whose price is text, using [`type()`](https://docs.python.org/3/library/functions.html#type) inside an [`if`](https://docs.python.org/3/tutorial/controlflow.html#if-statements). On a receipt of 1,200 lines, you could not read every type by eye.

**However:** It's worth noting that some developers frown upon using `type()` in favour of alternatives (such as `try`/`except`, which we'll be covering later). Having said that, `type()` is appropriate for debugging. It also helps when you need to tell a class apart from its subclasses, a more advanced topic we'll briefly mention next week.

````python
# Find the odd one out: Print only the records whose price is text.
for item, price in coffee_run:
    if type(price) == str:
        print(f"Text price: {item}, stored as '{price}'")
````

Were you confused by the `item, price` in the `for` loop above?

We use that because of this:

```
coffee_run = [
    ("flat white", 3.40),
    ...
```

`coffee_run` is a `list` of pairs, and Python calls each pair a `tuple`. The `item, price` gives the two parts of each pair their own names. Admittedly, it looks complicated and is probably quite confusing at first. But trust me: This is incredibly powerful, and you'll be glad that it exists ... not now, but soon!

## 🥉 Bronze: Worked answer

Here, a single loop does all the work.

It converts each price with [`float()`](https://docs.python.org/3/library/functions.html#float), adds it to a running total, and counts the items. `float()` leaves a price that is already a number unchanged (!), so it is safe on all twelve.

The running total is the pattern from section 6 of [the Demonstration's Python warm-up](./Demonstration.ipynb).

````python
# 🥉 Bronze: Total spend and average price per item, with one loop.
total = 0
count = 0
for item, price in coffee_run:
    total = total + float(price)   # float() turns the text "2.80" into 2.8
    count = count + 1

average = total / count

print(f"Total spend:      £{total:.2f}")
print(f"Average per item: £{average:.2f}  (divided over {count} items)")
````

**Is £3.07 exact?** ... Nope.

The true average is £3.0666…, and `:.2f` rounds it for display. That is **correct** for a message to the Partner!

### The verification steps

These two checks are what make the answer [Bronze](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-bronze). **Remember:** "*A number you have not checked is not ready to send.*"

- **The petty-cash check:** £40.00 went out and £3.20 came back, so £36.80 was spent. Your total must match it to the penny
- **The hand-check:** The first two items are £3.40 and £3.60. On a calculator, that is £7.00. The same loop over `coffee_run[:2]` must print £7.00 too

The cell below runs both. It compares the two totals after rounding each to the penny with [`round()`](https://docs.python.org/3/library/functions.html#round). Failure 2 in the [AI failure gallery](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-ai-failure-gallery) below shows why the rounding is there.

````python
# Verify before you send: The petty-cash check, then the hand-check.
petty_cash_spent = 40.00 - 3.20
print(f"Petty cash says spent: £{petty_cash_spent:.2f}")
print(f"Your loop says spent:  £{total:.2f}")
print(f"Reconciles to the penny: {round(total, 2) == round(petty_cash_spent, 2)}")

check_total = 0
for item, price in coffee_run[:2]:
    check_total = check_total + float(price)
print(f"Hand-check, first two items: £{check_total:.2f}  (calculator: £7.00)")
````

Both checks pass, so the numbers can go to the Partner. Remember: *A good check is one that **could have failed**.* This one could: A skipped item or a doubled item would break the check!

As an aside: Look up the `==` in `print(f"Reconciles to the penny: {round(total, 2) == round(petty_cash_spent, 2)}")`. It isn't the same as `=`. Again, it is confusing at first, but very (!) useful once you understand the distinction.

## 🥈 Silver: Worked answer

The [Associate](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-associate)'s overnight draft failed with a `TypeError`. [`sum()`](https://docs.python.org/3/library/functions.html#sum) cannot add the text `"2.80"` to a number (we talked about that in class).

Both fixes repair it, and both print the same total and average. So the [Silver](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-silver) question is not which fix works. It is which fix you would rather live with, and why.

**Fix A: Convert on read.** The cell below is the draft with one change: `float(price)` inside the [`.append()`](https://docs.python.org/3/tutorial/datastructures.html#more-on-lists). The average divides by [`len()`](https://docs.python.org/3/library/functions.html#len), the number of prices collected.

````python
# 🥈 Fix A: Convert on read. The draft, with float() where prices are collected.
prices = []
for item, price in coffee_run:
    prices.append(float(price))

# print(prices)  # See what happens ...

total = sum(prices)
average = total / len(prices)
print(f"Fix A: Total £{total:.2f}, average £{average:.2f}")
````

**Fix B: Clean the data once.** A cleaning step builds a repaired copy of the receipt, with every price a number.

In your own notebook it belongs **straight after** the data setup, and it can keep the name `coffee_run`. Here it builds `clean_run` instead, so the <!-- plain-language: everyday Gallery -->gallery below can still show the raw receipt.

````python
# 🥈 Fix B: Clean once. Build a repaired copy of the receipt, every price a number.
clean_run = []
for item, price in coffee_run:
    clean_run.append((item, float(price)))

print(f"Cleaned {len(clean_run)} records. The filter coffee is now {clean_run[8]}")
````

The cell below is the Associate's overnight draft, run on the cleaned list. Nothing in it changes except **the name of the list** it reads.

````python
# The Associate's overnight draft, unchanged except for the list it reads.
prices = []
for item, price in clean_run:
    prices.append(price)

total = sum(prices)
average = total / len(prices)
print(f"Fix B: Total £{total:.2f}, average £{average:.2f}")
````

## 🥇 Gold: One worked example

Any clearly stated rule is a valid [Gold](https://youtu.be/0O1URHDI07M) answer. That is, if you build it, print a table, and name the rule in one sentence.

This example follows the quest's first two ideas. Each item gains a third element: The colleague who ordered it. The names are invented, and the prices are already clean, as under Fix B.

````python
# 🥇 Gold: The receipt again, with who ordered each item (invented colleagues).
orders = [
    ("Amara", "flat white", 3.40),
    ("Ben", "latte", 3.60),
    ("Chloe", "cappuccino", 3.50),
    ("Dev", "espresso", 2.20),
    ("Grace", "hot chocolate", 3.20),
    ("Elena", "english breakfast tea", 2.40),
    ("Farid", "peppermint tea", 2.40),
    ("Amara", "almond croissant", 2.95),
    ("Grace", "filter coffee", 2.80),
    ("Dev", "banana bread", 2.75),
    ("Ben", "mocha", 3.80),
    ("Chloe", "oat flat white", 3.80),
]
people = ["Amara", "Ben", "Chloe", "Dev", "Elena", "Farid", "Grace"]

print(f"{len(orders)} orders from {len(people)} people")
````

The cell below compares two rules: An even split of the total, and everyone paying for what they ordered. A loop inside a loop totals each person's orders. The outer loop traces that through each person, and the inner loop adds up that person's items *(loop within loop)*.

The table lines up because of the f-string's [format codes](https://docs.python.org/3/library/string.html#format-specification-mini-language). `:<6` pads a name to six characters, `:>8.2f` right-aligns a figure, and `+` shows a plus or minus sign. In the last column, a plus means the even split charges that person more than their own order.

````python
# 🥇 Pay for what you ordered, against an even split of the total.
total = 0
for person, item, price in orders:
    total = total + price
even_share = total / len(people)

print("All amounts in pounds.")
print(f"{'Name':<6}  {'Ordered':>8}  {'Even split':>10}  {'Extra paid':>10}")
for name in people:
    ordered = 0
    for person, item, price in orders:
        if person == name:
            ordered = ordered + price
    extra = even_share - ordered
    print(f"{name:<6}  {ordered:>8.2f}  {even_share:>10.2f}  {extra:>+10.2f}")

shares_total = round(even_share, 2) * len(people)
print(f"\nEven split: {len(people)} shares of £{even_share:.2f} = £{shares_total:.2f}, "
      f"against £{total:.2f} spent")
````

Looking at the output, can you guess what `:<6` does? You see: It looks complicated, but it is really powerful ... and with your Associate, you won't have to remember it!

Let's look at the results: Elena and Farid each ordered one £2.40 tea, and an even split charges them £5.26. Each pays £2.86 towards someone else's order. Looks like Ben gains most: His latte and mocha cost £7.40 (crazy), and he pays £5.26.

The even split also fails to reconcile the overall result: Seven shares of £5.26 come to £36.82, which is **2p more** than what was spent. Not much, of course, but not a good look!

**The one-sentence rule:** Everyone pays for what they ordered, to the penny, so the shares add up to exactly £36.80. A stronger <!-- plain-language: everyday Gold -->Gold also names what it ignores, such as the intern's time spent fetching the order.

## 🚩 The bug and the fix ("Red Flags")

**The bug:** `average = total / 10` divides by a typed-in 10, and the receipt has 12 items. The summary reports an average of £3.68 per item. Your Bronze average is £3.07.

**In business terms:** Finance would believe the average item costs 61p more than it does. That overstates the price by a fifth. They might set next month's coffee budget from it, or decide that *Beans of Egham* is an expensive place. The comment `# average over the coffee run` makes the line look **deliberate**, so nobody questions it.

**Why a typed-in count is wrong even when it is right:** On a day with exactly ten items, the summary prints the correct average. That is luck, which we really don't want to rely on. The day the run has 14 items, it is wrong again, with no error and no warning.

So, count with `len()`, which always matches the data.

**What would have shown it:** Even a "plausibility range" is not enough here. £3.68 sits between the cheapest item (£2.20) and most expensive one (£3.80), so it looks reasonable. Two checks catch it. The first compares it with your Bronze average. The second multiplies back: An average times the number of items must give the total.

So far, so obvious ... but not always when we're trying to get this stuff to work.

So, let's look at the cell below that runs the multiply-back check on the Associate's summary.

````python
# The multiply-back check: Average times items must equal the total.
clean_prices = []
for item, price in coffee_run:
    clean_prices.append(float(price))
total = sum(clean_prices)
wrong_average = total / 10

print(f"£{wrong_average:.2f} × {len(clean_prices)} items = "
      f"£{wrong_average * len(clean_prices):.2f}")
print(f"Total actually spent:  £{total:.2f}")
````

Multiplied back, the summary's average implies £44.16 of coffee. Petty cash says £36.80.

**A directing prompt that fixes it:** While the quest's prompt works, a stronger prompt also asks for a check **you can see**:

*"This summary divides by a hard-coded 10, but the receipt has 12 items. Fix it so the divisor is always the number of items, using len(). Then print the item count next to the average, so a reader can check the division."*

The cell below is the fixed summary.

````python
# 🚩 Fixed summary: The divisor is the number of items, counted with len().
clean_prices = []
for item, price in coffee_run:
    clean_prices.append(float(price))

total = sum(clean_prices)
average = total / len(clean_prices)

print("=== Coffee run summary ===")
print(f"Items:            {len(clean_prices)}")
print(f"Total spend:      £{total:.2f}")
print(f"Average per item: £{average:.2f}")
````

The fixed summary prints the item count too.

A reader who sees "12" beside the average can check the division without opening the code. That's what you want!

### Failure 1: The loop that skips the awkward price

When asked to "make the total robust to bad data", the Associate adds a check. It skips any price that is not a number, and the cell runs cleanly.

Run it, then compare its figures with your Bronze answer.

````python
# Associate draft: 'Skip anything that is not a number, to be safe.'
total = 0
count = 0
for item, price in coffee_run:
    if type(price) == float:   # only add real numbers
        total = total + price
        count = count + 1

print(f"Total spend:      £{total:.2f}")
print(f"Average per item: £{total / count:.2f}")
````

**Why it is wrong:** The check does not repair the filter coffee's price. It drops the whole item. The total loses £2.80, and the average is taken over 11 items, not 12. Nothing warns you.

**In business terms:** Petty cash says £36.80 was spent, and this report says £34.00. Finance sees £2.80 missing from the tin. The intern who fetched the order looks like the obvious suspect, and did nothing wrong. The average, £3.09, is close enough to the truth to pass a glance, so it hides the gap.

### Failure 2: The check that cries wolf

Asked to "add a reconciliation check against petty cash", the Associate compares the loop total with `==`. The cell prints two identical figures, then a failure.

````python
# Associate draft: 'Check the total against petty cash.'
total = 0
for item, price in coffee_run:
    total = total + float(price)

petty_cash_spent = 40.00 - 3.20
print(f"Loop total: £{total:.2f}")
print(f"Petty cash: £{petty_cash_spent:.2f}")
if total == petty_cash_spent:
    print("Reconciles.")
else:
    print("Does NOT reconcile. Investigate!")
````

**Why it is wrong:** A computer stores most decimals, such as 3.40, very slightly inexactly. After twelve additions, the total sits a tiny fraction of a penny below 36.80. `==` demands an exact match, so it reports a gap too small to be money. The [Python tutorial on floating-point arithmetic](https://docs.python.org/3/tutorial/floatingpoint.html) explains why.

**In business terms:** Finance is told the coffee run does not reconcile. Someone spends an afternoon hunting a gap smaller than any coin. After two false alarms, people stop reading the check. Then a real £2.80 gap, like the one in Failure 1, goes unnoticed.

**The catch:** Print the gap itself, and compare money to the penny, as the Bronze check does. The cell below does both.

````python
# What would have shown it: Print the gap, then compare to the penny.
print(f"The gap: {total - petty_cash_spent}")
print(f"Reconciles to the penny: {round(total, 2) == round(petty_cash_spent, 2)}")
````

The `e-15` at the end of the gap means "move the decimal point 15 places left". The gap is less than a millionth of a millionth of a penny.

### Failure 3: The total from the chat window

Asked in the Gemini panel "what is the total of these twelve prices?", the Associate answers in prose, with no code. A typical reply looks like this:

> **Associate:** Adding up the twelve items gives a total of **£36.30**, so the average price per item is about **£3.03**.

**Why it is wrong:** A language model is poor at long mental arithmetic. It predicts likely-looking digits rather than adding them up. It sounds just as confident when it is wrong as when it is right. Here it is 50p short.

**In business terms:** A figure with no working behind it cannot be audited. If the Partner forwards £36.30, petty cash is 50p out, and nobody can say where the error sits. The Partner asked you to check the numbers, so the error is yours, not the tool's.

**The catch:** Ask the Associate for code, not for the answer, and run the code. A running-total trace shows every step, so any slip has an address. The cell below prints one.

````python
# A running-total trace: Every step visible, so a wrong total has an address.
running = 0
for item, price in coffee_run:
    running = running + float(price)
    print(f"{item:<22} £{float(price):.2f}   running total £{running:.2f}")
````
