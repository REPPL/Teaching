## 3. The kernel: The memory behind the page

Behind every notebook sits a **kernel**, a live Python process running inside your Codespace. When you run a cell, the code goes to the kernel. The kernel remembers everything you have created so far.

The page and the kernel are different things:

- The **page** (text, code, outputs) is saved to disk in your Codespace as you work, like any file in VS Code
- The **kernel** (the live variables) is temporary. Restart it, or let the Codespace stop from inactivity, and its memory is wiped. Your page is untouched

**Restarting the kernel** wipes its memory on purpose. Section 5 shows what that is for.

> ✏️ **Try it: Break your own memory:**
>
> 1. Click the **Restart** button (↻) in the toolbar above the notebook, or open the Command Palette with `Ctrl/Cmd+Shift+P` and run **Jupyter: Restart Kernel**. Confirm if asked
> 2. Run the cell below. It fails with the same NameError as before. `mystery_number` lived in the kernel, and the kernel is new
> 3. Your page still has every cell. Re-run the cell in section 2 that creates `mystery_number`, then run the cell below again. It works

````python
# After a restart this fails until you re-run the cell that creates mystery_number.
print("The kernel remembers mystery_number:", mystery_number)
````

## 5. Run all: The analyst's health check

You can build a notebook that works only because of the order you ran its cells. It breaks for anyone who runs it top to bottom, including you tomorrow.

Before you finish any working session, restart the kernel and run the whole notebook. Open the Command Palette and run **Jupyter: Restart Kernel and Run All Cells**. If the notebook runs cleanly from a blank kernel, top to bottom, your work is reproducible.

> ✏️ **Try it:** Run **Restart Kernel and Run All Cells** on this notebook. The run stops at the out-of-order cell in section 2. VS Code cancels the cells queued below the error. That stop is the health check doing its job. To finish the run:
>
> 1. Click the cell *after* the one that errored, the cell that creates `mystery_number`
> 2. Choose **Execute Cell and Below** from the dropdown next to its **Run** button
>
> Do not start from the failing cell itself. It fails again and cancels the queue. In your own notebooks, a stop during this check is your cue to fix the order before you finish. This teaching notebook keeps its one intended error. Yours should end up with none.

## 6. Where files live

Your Codespace has its own file system, the one the Explorer panel shows you. Code can write files there. The next cell uses two things you have not met yet. `open()` writes a text file, and the `os` module lists what is in the folder. [The official Python tutorial](https://docs.python.org/3/tutorial/inputoutput.html#reading-and-writing-files) covers reading and writing files. Run the cell and watch where the file lands.

````python
# Write a small file to the runtime's file system, then prove it exists.
with open("first_note.txt", "w") as f:
    f.write("Junior Analyst's note: files written here are temporary.\n")

import os
print("Files in the runtime's working folder:", os.listdir("."))
````

`first_note.txt` appears in the list, and in the Explorer panel on the left, in this same folder. That file lives in your Codespace's storage, not in GitHub. It survives you stopping and resuming your Codespace, but it is not backed up until you <!-- plain-language: everyday commit -->commit and push it.

Rules of survival:

- Your **notebook** is saved to your Codespace automatically as you work, like any file in VS Code. Watch the dot-to-tick indicator in the tab
- Any **file your code writes** lives alongside it. It is safe while your Codespace exists, but it is backed up only once you `git add`, `commit`, and `push` it. Section 8 shows how
