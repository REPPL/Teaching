## A little more arithmetic

Week 1 gave you the "basics": `+`, `-`, `*`, and `/`. Three more operators appear in business code and are really needed.

Unfortunately, they're not as clear as the ones above. Let's look at them in the next cell: Floor division (`//`), the remainder (`%`, also called modulo), and powers (`**`). The official Python tutorial covers them under [numbers](https://docs.python.org/3/tutorial/introduction.html#numbers).

````python
print(10 / 3)    # division always returns a decimal (a float)
print(10 // 3)   # floor division (NEW): Whole-number part only
print(10 % 3)    # modulo (NEW): The remainder
print(2 ** 10)   # exponentiation - NEW: 2 to the power of 10
````

Floor division and remainder answer two questions a stockroom asks: How many full boxes does a delivery fill, and how many cups are left over?

````python
# Where you actually meet these: Packing 130 cups into boxes of 12.
cups = 130
print(f"Full boxes:    {cups // 12}")
print(f"Cups left over: {cups % 12}")
````

## Multi-way decisions with `elif`

Week 1's `if`/`else` makes a two-way <!-- plain-language: everyday decision -->decision.

`elif` ("else if") adds more branches, such as morning, lunchtime, or afternoon ([`if` statements](https://docs.python.org/3/tutorial/controlflow.html#if-statements) in the official tutorial). Python checks the conditions **top to bottom, and the first true one wins**, so order matters.

````python
hour = 15   # try 8, 12, and 15

if hour < 11:
    slot = "morning"
elif hour < 14:  # !!!
    slot = "lunchtime"
else:
    slot = "afternoon"

print(f"A sale at {hour}:00 lands in the {slot} slot")
````

There is no `hour >= 11` in the `elif`. By the time Python reaches the `elif`, `hour < 11` has already failed, so the `elif` does not need to repeat that test. This pattern (a value in, a category out) appears in this week's [quest](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-quest).

## Functions: Code you can reuse

A **function** is a named, reusable block of code. Inputs go in as *parameters* and a result comes out through `return` ([defining functions](https://docs.python.org/3/tutorial/controlflow.html#defining-functions) in the official tutorial). A function gives a business rule one name and one place in your code, so you can test the rule on its own.

```python
def function_name(parameter):
    # work happens here
    return result
```

````python
def add_vat(price):
    # Add 20% VAT to a price and return the total.
    return price * 1.20

final_price = add_vat(100)
print(f"Price with VAT: £{final_price:.2f}")
print(f"Works for any price: £{add_vat(3.40):.2f}")
````

The next cell puts an `elif` chain inside a function called `loyalty_tier`, so you can call the rule with any spend.

````python
# A function wrapping an elif chain: The rule now has a name and is testable.
def loyalty_tier(monthly_spend):
    if monthly_spend >= 60:
        return "Gold"
    elif monthly_spend >= 30:
        return "Silver"
    elif monthly_spend >= 10:
        return "Bronze"
    else:
        return "New customer"

for spend in [75, 42, 15, 3]:
    print(f"£{spend} per month: {loyalty_tier(spend)}")
````

Parameters can carry **default values** ([default argument values](https://docs.python.org/3/tutorial/controlflow.html#default-argument-values) in the official tutorial). You give a value only when you want to change the default. At a coffee cart, the default is one cup.

````python
# Parameters can have DEFAULT values: Callers can omit them.
def receipt_line(item, cups=1):
    return f"{cups} x {item}"

print(receipt_line("flat white"))          # uses the default
print(receipt_line("latte", 3))            # overrides it
````

`return` can hand back more than one value. You store them in separate variables on one line.

````python
# A function can return MORE THAN ONE value.
def describe_takings(amounts):
    total = sum(amounts)
    average = total / len(amounts)
    return total, average

week = [412, 385, 447, 391, 508]
total, average = describe_takings(week)
print(f"Total £{total}, average £{average:.2f} per day")
````

`while` repeats for as long as a condition stays true. How many flat whites does 800ml of milk make? The loop finds out.

````python
# while repeats as long as a condition holds - here, until the milk runs out.
milk_ml = 800
cups_made = 0

while milk_ml >= 180:      # a flat white needs 180ml
    milk_ml = milk_ml - 180
    cups_made = cups_made + 1

print(f"Made {cups_made} flat whites; {milk_ml}ml of milk left")
````

## String methods

Text data arrives messy. It has stray spaces, inconsistent capitals, and several values in one string.

Strings carry built-in **methods** (these are functions attached to the value, which you call with a dot) for cleaning text up.

The official docs list [every string method](https://docs.python.org/3/library/stdtypes.html#string-methods). A string method **returns a new string** and never changes the original.

````python
order = "  Flat White, LARGE, 2  "

print(order.strip())            # remove whitespace at both ends
print(order.lower())            # all lowercase
print(order.upper())            # all uppercase
print(order.replace("2", "3"))  # replace text
print(order.strip().split(","))  # split into a LIST of parts: Very useful
````

Because every string method returns a new string, methods **chain**: The output of one is the input of the next, read left to right.

````python
# Chaining methods: Clean each part of a messy order line in one pass.
order = "  Flat White, LARGE, 2  "
parts = order.strip().split(",")

for part in parts:
    print(part.strip().lower())
````

## Lists: New techniques

You can create lists and loop over them. You can also pull an item out by position (**indexing**, from 0), take a run of items (**slicing**), and sort. The tutorial's [introduction to lists](https://docs.python.org/3/tutorial/introduction.html#lists) covers indexing and slicing.

````python
daily_takings = [412, 385, 447, 391, 508]

print(daily_takings[0])     # first item (counting starts at 0)
print(daily_takings[-1])    # last item (negative counts from the end)
print(daily_takings[1:3])   # slice: Items 1 and 2 (stops BEFORE 3)
print(daily_takings[:2])    # first two items
````

Lists grow and shrink in place. You met `.append()` in Week 1. [`.remove()`](https://docs.python.org/3/tutorial/datastructures.html#more-on-lists) takes an item out by value.

````python
# Growing and shrinking a list
specials = ["gingerbread latte"]
specials.append("iced mocha")       # add to the end
specials.append("turmeric latte")
specials.remove("turmeric latte")   # remove by value (it did not sell)
print(specials)
````

Sorting comes in two forms that <!-- plain-language: everyday Analyst -->analysts and Associates often confuse. `sorted()` returns a **new** list and leaves the original untouched. `.sort()` rearranges the list itself ([Sorting HOW TO](https://docs.python.org/3/howto/sorting.html) in the official docs).

````python
takings = [412, 385, 447, 391, 508]

print(f"sorted() copy:  {sorted(takings)}")          # returns a NEW sorted list
print(f"original intact: {takings}")

takings.sort(reverse=True)                            # .sort() changes the list itself
print(f"after .sort():  {takings}")
````

In the build-a-new-list pattern you start with an empty list, loop over the old list, and `.append()` a transformed value each time. Here the loop takes the card processor's cut from every day's takings. Arithmetic on floats leaves long decimal tails. The `:.2f` formatting from Week 1 hides them in the printed output.

````python
# Building a new list from an old one: A pattern you will use constantly.
takings = [412, 385, 447, 391, 508]
with_card_fees = []

for amount in takings:
    with_card_fees.append(amount * 0.986)   # card processor takes 1.4%

for amount in with_card_fees:
    print(f"£{amount:.2f}")
````

## Dictionaries: Labelled data

A list stores values *by position*. A **dictionary** stores values *by name*, as key-value pairs ([dictionaries](https://docs.python.org/3/tutorial/datastructures.html#dictionaries) in the official tutorial). A price list, a customer record, and a till transaction all fit this shape.

````python
menu = {
    "flat white": 3.40,
    "latte": 3.60,
    "espresso": 2.20,
}

print(menu["latte"])            # look up a value by its key
menu["mocha"] = 3.80            # add a new pair
menu["espresso"] = 2.30         # update an existing one
print(menu)
````

Looking up a missing key with square brackets crashes the cell. [`.get()`](https://docs.python.org/3/library/stdtypes.html#dict.get) returns a default instead. With real data, the data decides which keys exist, not you.

````python
# Safe lookups with .get(): A missing key gives a default instead of an error.
menu = {"flat white": 3.40, "latte": 3.60, "espresso": 2.30}

print(menu.get("latte", "not on the menu"))
print(menu.get("babyccino", "not on the menu"))   # menu["babyccino"] would CRASH
````

You can loop over a dictionary too. [`.items()`](https://docs.python.org/3/library/stdtypes.html#dict.items) hands you each key and value as a pair, ready to use as two loop variables.

````python
# Looping over a dictionary: .items() gives you each key and value together.
menu = {"flat white": 3.40, "latte": 3.60, "espresso": 2.30}

for item, price in menu.items():
    print(f"{item}: £{price:.2f}")
````

The **counting** pattern is this week's main analytics tool. Loop over the raw orders and let `.get(drink, 0)` start each drink's tally at zero the first time it appears.

````python
# Counting with a dictionary: THE analytics pattern of the week.
orders = ["latte", "flat white", "latte", "espresso", "latte", "flat white"]

tally = {}
for drink in orders:
    tally[drink] = tally.get(drink, 0) + 1   # start at 0 if unseen, then add 1

print(tally)
````

One step up is a **list of dictionaries**, one dictionary per record. Much real business data arrives in this shape, and so does this week's quest data.

````python
# A LIST OF DICTIONARIES: One dictionary per record.
# This is the exact shape of the quest's till data.
transactions = [
    {"item": "latte", "cups": 2, "price": 3.60},
    {"item": "espresso", "cups": 1, "price": 2.30},
    {"item": "latte", "cups": 1, "price": 3.60},
]

revenue = 0
for sale in transactions:
    revenue = revenue + sale["cups"] * sale["price"]

print(f"Revenue from {len(transactions)} transactions: £{revenue:.2f}")
````

## File handling: Write it, read it back

Data arrives in files, and analyses end up in files. The `with open(...)` pattern opens a file and **closes it automatically** when the block ends, even if something goes wrong halfway. The official tutorial covers [reading and writing files](https://docs.python.org/3/tutorial/inputoutput.html#reading-and-writing-files). The mode string says what you intend. Use `"r"` to read and `"w"` to write. Opening a file in `"w"` mode **replaces** its contents.

The data-setup cell at the top creates `assets/data/week02/takings.txt`. The next cell reads it line by line.

````python
# Reading a file, line by line.
with open("assets/data/week02/takings.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    print(line.strip())    # .strip() removes the invisible newline at the end
````

Printing lines is not analysis. **Parsing** turns a file into data: Split each line into its parts, then convert the text `"412"` into a number you can add up.

````python
# Parsing: Turning each line of text into values you can compute with.
total = 0
with open("assets/data/week02/takings.txt", "r") as file:
    for line in file:
        parts = line.strip().split(": £")     # "Monday: £412" -> ["Monday", "412"]
        day = parts[0]
        amount = int(parts[1])                # text "412" -> number 412
        total = total + amount
        print(f"{day}: £{amount}")

print(f"\nWeek total: £{total}")
````

Now write a file. The next cell opens it in `"w"` mode, writes a short summary, then reads it straight back to prove it saved.

````python
# Writing a new file: A one-line summary a manager could open in Notepad.
with open("assets/data/week02/summary.txt", "w") as file:
    file.write("Beans of Egham - weekly takings summary\n")
    file.write(f"Week total: £{total}\n")
    file.write(f"Daily average: £{total / 5:.2f}\n")

# Prove it worked: Read it straight back.
with open("assets/data/week02/summary.txt", "r") as file:
    print(file.read())
````
