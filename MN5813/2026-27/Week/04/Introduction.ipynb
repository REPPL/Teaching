## Before and after pandas

In [Quest](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-quest) 01, you totalled twelve coffees with a loop. Run the cells below. They do the same job twice, once with a loop and once with pandas.

This cell creates a short list of sales directly in Python, with no file. The Demonstration and the Quest generate their own files. Run it ... run it now!

*Note:* Remember what *data type* the variable `sales_datat` below is?

```python
# --- Data setup: Run this cell first (runs anywhere, no download needed) ---
# A small transaction list, typed in as plain Python (no files this time;
# the Demonstration and Quest generate real files with the module seed, 5813).
sales_data = [
    {"date": "2026-01-05", "product": "Table Lamp", "store": "Egham", "amount": 34.00},
    {"date": "2026-01-05", "product": "Cushion", "store": "Staines", "amount": 12.50},
    {"date": "2026-01-06", "product": "Mirror", "store": "Egham", "amount": 45.00},
    {"date": "2026-01-07", "product": "Throw", "store": "Windsor", "amount": 24.00},
    {"date": "2026-01-07", "product": "Candle Set", "store": "Staines", "amount": 9.75},
    {"date": "2026-01-08", "product": "Vase", "store": "Egham", "amount": 16.50},
]
print(f"Created sales_data: {len(sales_data)} transactions")
```

The cell below totals the sales the Week 1 way. It uses a `loop` (remember?) with running totals, the pattern you built in Quest 01.

```python
# The Week 1 way: Loops and running totals. Correct, but it scales in EFFORT.
total = 0
store_totals = {}
for sale in sales_data:  # the loop
    total += sale["amount"]
    store_totals[sale["store"]] = store_totals.get(sale["store"], 0) + sale["amount"]

print(f"Total sales: £{total:.2f}")
print(f"By store:   {store_totals}")
```

The pandas way answers the same questions. [`pd.DataFrame`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html) turns the list into a table, and [`groupby`](https://pandas.pydata.org/docs/user_guide/groupby.html) totals the sales by store in one line.

```python
# The pandas way: The same questions become one-liners that scale to millions of rows.
# Convenient: We reserve the name 'pd' for pandas, so we can type less.
import pandas as pd

df = pd.DataFrame(sales_data) # pd. means: From pandas, use DataFrame()

print(f"Total sales: £{df['amount'].sum():.2f}")
print(df.groupby("store")["amount"].sum())
```

The loop is correct, but it is more work. The pandas version is shorter. It is also easier to read and check when your Associate wrote it.

## Key concepts/ideas

### 1. `Series`

A [`Series`](https://pandas.pydata.org/docs/reference/api/pandas.Series.html) is **one column** of data, ordered like a list and labelled like a dictionary. Every column of a `DataFrame` is a `Series`. Run the cell to build a `Series` of prices and print its average.

```python
prices = pd.Series([34.00, 12.50, 45.00, 24.00],
                   index=["Table Lamp", "Cushion", "Mirror", "Throw"])

print(prices)
print(f"\nAverage price: £{prices.mean():.2f}")
```

### 2. `DataFrames`

Each column of a `DataFrame` can contain a different kind of value, such as text, numbers, or dates. Every step is written as code, so you can run it again and check it. Run the cell to build one from the sales list and show its first rows with [`head()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.head.html).

```python
df = pd.DataFrame(sales_data)
df.head()
```

### 3. Filtering

A **condition** turns a business question (*which sales were over £20?*) into code. Run the cell to keep only the rows where `amount` is over 20. This is called [boolean indexing](https://pandas.pydata.org/docs/user_guide/indexing.html#boolean-indexing).

```python
big_sales = df[df["amount"] > 20]
print(big_sales)
```

### 4. Groupby

`groupby` splits the rows into groups, applies a calculation to each group, and combines the results. Revenue *by store*, medals *by country*, and complaints *by month* are all `groupby()` questions. Run the cell to get the total and the count of sales for each store with [`agg`](https://pandas.pydata.org/docs/user_guide/groupby.html#aggregation).

```python
print(df.groupby("store")["amount"].agg(["sum", "count"]))
```
