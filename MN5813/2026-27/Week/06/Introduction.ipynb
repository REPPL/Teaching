## The encoding ladder

In what's widely considered a "classic" study, Cleveland and McGill (1984) ranked how **accurately** people read values from each visual encoding. The [DataViz design guide](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/dataviz-design-guide.md), section "2. T: Task", explains some of its implications:

1. Typically considered "best": **Position on a common scale** (dots, or bar ends, against one axis)
2. **Position on separate scales** (the same measure in side-by-side panels whose axes do not line up)
3. **Length, direction, and angle** (a bar with no common baseline, the slope of a line, a pie slice)
4. **Area** (a bubble or circle)
5. **Volume and curvature** (a 3D block, a curve)
6. **Shading** (a darker or lighter cell of one colour)

Encodings at the top let a reader *measure* with their eyes. Encodings at the bottom only show that one thing is bigger than another. When the reader needs to compare values precisely, use **position on a common scale**: Dots, or bars that share a baseline.

This ranking is why a bar chart is easier to read than a pie chart *(that's why we don't like them)*. A reader compares bars by where they end, and pie slices by their angle.

### Testing how you read each encoding

The cell below draws the same numbers as position, length, angle, area, and shading, in order down the ladder.

Run the cell. Then try to answer "how much bigger is East than Central?" from each panel in turn, left to right. Notice where your answer stops being a number and becomes a guess.

````python
#| fig-alt: "Five small charts show the same five sales figures, one per visual encoding. From left to right, they are dots on a position scale, bars on offset baselines, a pie, circles of different areas, and shaded squares. The values are easy to read from position, harder from length and angle, and hard from area and shade."
values, labels = sales.to_numpy(), list(sales.index)
x = np.arange(5)

fig, axes = plt.subplots(1, 5, figsize=(13, 3.2))
fig.suptitle("The same five sales figures, drawn with five different encodings", y=1.08)

# 1 POSITION: Dots on a common scale
ax = axes[0]; ax.set_title("Position")
ax.scatter(values, x, color="#0072B2", zorder=3)
ax.set_yticks(x, labels); ax.set_xlim(0, 47)
ax.grid(axis="x", alpha=0.3); ax.spines[["top", "right"]].set_visible(False)

# 2 LENGTH: Bars on offset baselines, so only their length shows the value
ax = axes[1]; ax.set_title("Length")
ax.bar(x, values, bottom=[3, 11, 0, 8, 15], color="#0072B2", width=0.55)
ax.set_xticks(x, labels, rotation=90); ax.set_yticks([])
for s in ax.spines.values(): s.set_visible(False)

# 3 ANGLE: A pie
ax = axes[2]; ax.set_title("Angle")
ax.pie(values, labels=labels, textprops={"fontsize": 8})

# 4 AREA: Circles, area proportional to value
ax = axes[3]; ax.set_title("Area")
ax.scatter(x, np.zeros(5), s=values * 40, color="#0072B2", alpha=0.8)
ax.set_xticks(x, labels, rotation=90); ax.set_yticks([]); ax.set_ylim(-1, 1)
for s in ax.spines.values(): s.set_visible(False)

# 5 SHADING: Shade proportional to value
ax = axes[4]; ax.set_title("Shading")
ax.imshow(values.reshape(1, -1), cmap="Blues", aspect="auto", vmin=0, vmax=47)
ax.set_xticks(x, labels, rotation=90); ax.set_yticks([])

plt.tight_layout()
plt.show()
````

**What to notice.** From the *position* panel you can read values off the scale: East about 26 and Central about 9, so roughly three times as much. From *length* without a common baseline, and from *angle*, you are estimating. The ladder ranks these two together. From *area* you are guessing. From *shading* you can say little more than "East is darker than Central".
