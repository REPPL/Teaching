## Polishing your brief: Testing your rationale

Polishing your [brief](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-brief) tests your **angle** (rationale), the question you developed from your brief's starting question. A viable rationale passes these tests:

- **It uses your brief's primary table, slice, and primary measure**: The [briefs page](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Assessment/individual-briefs.md) sets all three
- **It is about the right population**: You take a <!-- plain-language: everyday decision -->decision on each challenge in section "The challenges (well, some of them!)" of the [primer](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Assessment/abcd-record.md)
- **It avoids the poor choices**: Section "2. Dataset" of the [individual report brief](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Assessment/individual-report.md) lists them

A finding critical of the project passes too, as section "3. The client" of the [individual report brief](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Assessment/individual-report.md) explains.

The code cells below are a **thirty-second check** on the primary table your brief mentions. The first cell builds a demonstration table. You need to run it first *(obviously)* ...

````python
# --- Data setup: Run this cell first (runs anywhere, no download needed) ---
import numpy as np
import pandas as pd

rng = np.random.default_rng(5813)  # module seed: Do not change

# A stand-in "demonstration table" so the check below has something to inspect.
n = 1200
candidate = pd.DataFrame({
    "order_id":   range(10001, 10001 + n),
    "region":     rng.choice(["London", "Midlands", "North", "Scotland", "Wales"], size=n),
    "channel":    rng.choice(["Online", "In store"], size=n, p=[0.7, 0.3]),
    "units":      rng.integers(1, 12, size=n),
    "unit_price": rng.uniform(2.5, 40.0, size=n).round(2),
})
candidate["revenue"] = (candidate["units"] * candidate["unit_price"]).round(2)
print(f"Created demonstration table in memory: {len(candidate)} rows x {candidate.shape[1]} columns")
````

The next cell runs the check. It sorts your columns into numerical and everything else with pandas' [`select_dtypes`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.select_dtypes.html). It counts rows and missing values. It prints a verdict against the rules in section "2. Dataset" of the [individual report brief](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Assessment/individual-report.md). To check your own table, replace `df = candidate` with the line the comment gives.

````python
# The viability check. For your own data, replace the first line with:
#   df = pd.read_csv("your_candidate.csv")
df = candidate

numeric = df.select_dtypes(include="number").columns.tolist()
non_numeric = [c for c in df.columns if c not in numeric]

print(f"Rows: {len(df):,}   (rule: at least 1,000)")
print(f"Numerical variables ({len(numeric)}): {numeric}")
print(f"Categorical/other variables ({len(non_numeric)}): {non_numeric}")
print(f"Missing values: {int(df.isna().sum().sum()):,}")

rows_ok = len(df) >= 1000
mix_ok = len(numeric) >= 2 and len(non_numeric) >= 2
verdict = (
    "meets the rules. Now check your angle against the tests above"
    if (rows_ok and mix_ok)
    else "does not yet meet the rules. Talk to me before you choose it"
)
print(f"\nVerdict: {verdict}")
````

⚠️ **Remember: A table can pass the check and still be a poor choice.** The check counts `order_id` as a "numerical variable", but it is an identifier. Your angle still has to pass the tests above. I judge it against your brief.
